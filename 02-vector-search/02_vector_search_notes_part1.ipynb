# Vector Search with Qdrant

## What is Vector Search and How It Works

### Traditional Keyword Search vs Vector Search
Traditional keyword search operates through a fundamentally straightforward but limited mechanism that relies on matching exact keywords within documents. When a user searches for a specific term like "bats," the system will only return documents that explicitly contain that exact word, making it a purely text-based matching system focused on particular keywords. This approach inherently cannot understand semantic meaning or context, limiting its effectiveness when users express their search intent using terminology different from what appears in the target documents.

Vector search operates on a semantic level, where the system focuses on understanding meanings rather than simply matching keywords. This advanced approach extends beyond traditional text-only searches to encompass multiple types of media, including text documents, images, videos, and audio files. Perhaps most importantly, vector search demonstrates the capability to match different words that share the same meaning, enabling semantic connections such as linking "bat" with "flying mouse" despite their lexical differences. The foundation of this capability lies in the use of vectorized representations of data, where the system processes numerical vector representations rather than working directly with raw text or media content.

### How Vector Search Works
The vectorization process converts diverse content types, including text, images, audio, or video, into numerical vector representations. Through this transformation, each piece of content becomes a vector positioned within a high-dimensional mathematical space, where the numerical values capture and encode the semantic essence of the original content. This mathematical representation enables computational operations that would be impossible with raw content formats.

Similarity measurement in vector search systems operates through numerical calculations that determine the relationships between different pieces of content. The system employs various similarity metrics to calculate the closeness between vectors within the high-dimensional space, with the fundamental principle that vectors positioned closer together represent content that is more semantically similar.

Semantic understanding enables the system to identify and retrieve relevant content even when users employ different vocabulary or phrasing than what appears in the target documents. The system's ability to understand conceptual relationships and semantic connections makes it far more intuitive and effective for users who may not know the precise terminology used in the documents they seek.

## Scaling Vector Search and Offline/Online Workflow
### The Scalability Challenge (Small Scale vs Large Scale)
Working with small datasets like our 1,000-document example allows for simple vector comparisons, but enterprise applications often deal with billions of vectors. Direct vector comparison becomes computationally expensive and impractical as datasets grow larger. Dedicated vector search engines with optimized indexing become essential for maintaining performance at scale.

### Two-Stage Workflow Architecture
**Stage 1: Offline Processing (Indexing)**: The offline stage involves taking all answer documents and processing them through an embedding model to create vector representations. These vectors are then stored in the Qdrant database along with an optimized index structure for scalable search operations. This process happens once for all documents and can be resource-intensive since it runs offline without time constraints.

**Stage 2: Online Processing (Query Time)**: The online stage processes incoming user questions in real-time by converting them into vectors using the same embedding model. The system then compares the query vector against stored answer vectors to find the most semantically similar matches. This stage must be optimized for speed and low latency to provide a good user experience.

Understanding how embeddings are compared is essential because the similarity metric determines how search results are ranked and which documents are considered relevant. The metric must align with how the embedding model was trained and its intended use case. Most embedding models are designed to work with cosine similarity, making it the standard choice for vector search applications.

Cosine similarity measures the cosine of the angle between two vectors in high-dimensional space, producing values ranging from -1 to +1. When vectors point in similar directions (small angles), they have high similarity scores close to +1, indicating semantically similar content. Vectors pointing in opposite directions (large angles) have low or negative scores, indicating different or contradictory meanings.

Now we’re ready to configure and use Qdrant for semantic search. To fully understand what’s happening, here’s a quick overview of Qdrant’s core terminology:
- **Points** are the central entity Qdrant works with. A point is a record consisting of an **ID**, a **vector**, and an optional **payload**. Points serve as the fundamental data records within the Qdrant vector database, containing all the information needed for semantic search operations. Each point consists of three essential components: a unique identifier (ID), the embedding vector that represents your data in mathematical space, and an optional payload containing metadata. The payload component stores valuable additional information such as the original text content, course names, and section details that enhance search results and enable filtering capabilities.
- A **collection** is a named set of points (i.e., vectors with optional payloads) that you can search within.
***Think of it as the container for your vector search solution, a single business problem solved***.
Qdrant supports different types of vectors to enable different modes of data exploration and search (dense, sparse, multivectors, and named vectors).

In this example, we’ll use the most common type, **dense vectors**.

Embeddings capture the semantic essence of the data, while the **payload** holds structured metadata.
This metadata becomes especially useful when applying filters or sorting during search. **Qdrant's payloads** can hold structured data like booleans, keywords, geo-locations, arrays, and nested objects.

## Step 4: Create a Collection
When creating a [collection](https://qdrant.tech/documentation/concepts/collections/), we need to specify:
- Name: A unique identifier for the collection.
- Vector Configuration:
    - Size: The dimensionality of the vectors. The model's output dimensions (512 for our selected model)
    - Distance Metric: The method used to measure similarity between vectors.
There are additional parameters you can explore in our [documentation](https://qdrant.tech/documentation/concepts/collections/#create-a-collection). Moreover, you can configure other vector types in Qdrant beyond typical dense embeddings (i.e., for hybrid search). However, for this example, the simplest default configuration is sufficient.

````python
# Define the collection name
collection_name = "zoomcamp-rag"
EMBEDDING_DIMENSIONALITY = 512

# Create the collection with specified vector parameters
client.create_collection(
    collection_name=collection_name,
    vectors_config=models.VectorParams(
        size=EMBEDDING_DIMENSIONALITY,  # Dimensionality of the vectors
        distance=models.Distance.COSINE  # Distance metric for similarity search
    )
)
````

Once created, the collection becomes visible in both your Python client environment and the Qdrant Web UI dashboard. The dashboard displays configuration details including dimensionality and distance metrics, while initially showing an empty collection awaiting data points. This verification step ensures that your collection is properly configured before proceeding with data upload operations.

## Step 5: Create, Embed & Insert Points into the Collection
[Points](https://qdrant.tech/documentation/concepts/points/#points) are the core data entities in Qdrant. Each point consists of:
1. **ID**. A unique identifier. Qdrant supports both 64-bit unsigned integers and UUIDs.
2. **Vector**. The embedding that represents the data point in the vector space.
3. **Payload** (optional). Additional metadata as key-value pairs.

The point creation process begins with ID assignment, typically using simple incremental integers for demonstration purposes, though production systems often employ UUID schemes for better scalability. Vector generation leverages FastEmbed's seamless integration with Qdrant, allowing you to specify the text content and model name while the system handles embedding generation automatically. The payload structure includes essential metadata fields such as the original text for human readability, section names for filtering by course sections, and course names for broader filtering capabilities.

````python
points = []
id = 0

for course in documents_raw:
    for doc in course['documents']:

        point = models.PointStruct(
            id=id,
            vector=models.Document(text=doc['text'], model=model_handle), #embed text locally with "jinaai/jina-embeddings-v2-small-en" from FastEmbed
            payload={
                "text": doc['text'],
                "section": doc['section'],
                "course": course['course']
            } #save all needed metadata fields
        )
        points.append(point)

        id += 1
````

Now we’re going to embed and upload points to our collection.

The upload process involves three main stages: model download and setup, embedding generation, and database insertion. During the first-time setup, FastEmbed automatically downloads and caches the selected model locally (path defaults to `os.path.join(tempfile.gettempdir(), "fastembed_cache"))` for future use. The embedding generation stage converts each answer into vector representations within a consistent vector space, ensuring all documents are comparable. Finally, the database insertion uploads points sequentially to the Qdrant collection, with each point containing its ID, vector representation, and associated payload metadata.

````python
client.upsert(
    collection_name=collection_name,
    points=points
)
````

The speed of upsert mainly depends on the time spent on local inference. ## Step 6: Running a Similarity Search
Now, let’s find the most similar `text` vector in Qdrant to a given query embedding - the most relevant answer to a given question.

### How Similarity Search Works
1. Qdrant compares the query vector to stored vectors (based on a vector index) using the distance metric defined when creating the collection.
2. The closest matches are returned, ranked by similarity.
> Vector index is built for **approximate** nearest neighbor (ANN) search, making large-scale vector search feasible.

If you'd like to dive into our choice of vector index for vector search, check our article ["What is a vector database"](https://qdrant.tech/articles/what-is-a-vector-database/), or, for a more technical deep dive, our article on Filterable [Hierarchical Navigable Small World](https://qdrant.tech/articles/filtrable-hnsw/).

Let's define a search function:

````python
def search(query, limit=1):

    results = client.query_points(
        collection_name=collection_name,
        query=models.Document( #embed the query text locally with "jinaai/jina-embeddings-v2-small-en"
            text=query,
            model=model_handle 
        ),
        limit=limit, # top closest matches
        with_payload=True #to get metadata in the results
    )

    return results
````

## Step 7: Running a Similarity Search with Filters
We can refine our search using metadata filters.
> Qdrant’s custom vector index implementation, Filterable HNSW, allows for precise and scalable vector search with filtering conditions.

For example, we can search for an answer to a question related to a specific course from the three available in the dataset.
Using a `must` filter ensures that all specified conditions are met for a data point to be included in the search results.
> Qdrant also supports other filter types such as `should`, `must_not`, `range`, and more. For a full overview, check our [Filtering Guide](https://qdrant.tech/articles/vector-search-filtering/)

To enable efficient filtering, we need to turn on [indexing of payload fields](https://qdrant.tech/documentation/concepts/indexing/#payload-index).

````python
client.create_payload_index(
    collection_name=collection_name,
    field_name="course",
    field_schema="keyword" # exact matching on string metadata fields
)
````

````python
# Now let's update our search function

def search_in_course(query, course="mlops-zoomcamp", limit=1):

    results = client.query_points(
        collection_name=collection_name,
        query=models.Document( #embed the query text locally with "jinaai/jina-embeddings-v2-small-en"
            text=query,
            model=model_handle
        ),
        query_filter=models.Filter( # filter by course name
            must=[
                models.FieldCondition(
                    key="course",
                    match=models.MatchValue(value=course)
                )
            ]
        ),
        limit=limit, # top closest matches
        with_payload=True #to get metadata in the results
    )

    return results
````

# RAG Integration with Vector Search
## RAG Architecture Overview: The Three-Step RAG Process
RAG (Retrieval-Augmented Generation) follows a modular three-step process that provides flexibility and maintainability in implementation. The first step involves searching for relevant documents using any replaceable search component, followed by building a context-rich prompt for the language model. The final step sends this constructed prompt to the LLM for generating the response, creating a clear separation of concerns that allows easy swapping between different search methods.

This modular design enables developers to experiment with various search implementations while maintaining the same overall architecture. The original implementation typically used traditional keyword-based search methods like Elasticsearch, which relied on exact word matching and had limited semantic understanding capabilities.

Below code is from our previous module.

````python
import requests 

docs_url = 'https://github.com/alexeygrigorev/llm-rag-workshop/raw/main/notebooks/documents.json'
docs_response = requests.get(docs_url)
documents_raw = docs_response.json()

documents = []

for course in documents_raw:
    course_name = course['course']

    for doc in course['documents']:
        doc['course'] = course_name
        documents.append(doc)

documents[2]
````

````python
import minsearch
from qdrant_client import QdrantClient
from openai import OpenAI

openai_client = OpenAI()

index = minsearch.Index(
    text_fields=["question", "text", "section"],
    keyword_fields=["course"]
)
index.fit(documents)
````

````python
def search(query):
    boost = {'question': 3.0, 'section': 0.5}
    results = index.search(
        query=query,
        filter_dict={'course': 'data-engineering-zoomcamp'},
        boost_dict=boost,
        num_results=5
    )
    return results

def build_prompt(query, search_results):
    prompt_template = """
You're a course teaching assistant. Answer the QUESTION based on the CONTEXT from the FAQ database.
Use only the facts from the CONTEXT when answering the QUESTION.

QUESTION: {question}

CONTEXT: 
{context}
""".strip()
    context = ""
    for doc in search_results:
        context = context + f"section: {doc['section']}\nquestion: {doc['question']}\nanswer: {doc['text']}\n\n"
    prompt = prompt_template.format(question=query, context=context).strip()
    return prompt

def llm(prompt):
    response = openai_client.chat.completions.create(
        model='gpt-4o-mini',
        messages=[{"role": "user", "content": prompt}]
    )
    return response.choices[0].message.content

def rag(query):
    search_results = search(query)
    prompt = build_prompt(query, search_results)
    answer = llm(prompt)
    return answer

rag('how do I run kafka?')
````

````python
rag('the course has already started, can I still enroll?')
````

### Adding Filters and Creating Indexes for Faster Search
**Enhanced Collection Configuration**
Creating an optimized collection involves setting up proper vector configurations and implementing smart naming strategies to avoid conflicts. The collection configuration specifies 512 dimensions and cosine similarity distance metrics to match the chosen embedding model. When working with existing collections, developers can delete and recreate them when schema changes or complete dataset refreshes are needed.

````python
# If collection already exists (collection management)
qd_client.delete_collection(collection_name=collection_name)

qd_client.create_collection(
    collection_name=collection_name,
    vectors_config=models.VectorParams(
        size=EMBEDDING_DIMENSIONALITY,
        distance=models.Distance.COSINE
    )
)
````

**Performance Optimization with Indexes**

Creating payload indexes for frequently filtered fields like course names dramatically improves query performance, especially in large collections. These indexes become essential for production environments where filtering operations are common and response times matter. The indexing process requires initial time investment but provides significant long-term performance benefits for scalable applications.

````python
qd_client.create_payload_index(
    collection_name=collection_name,
    field_name="course",
    field_schema="keyword"
)
````

The enhanced approach combines question and answer text during vectorization to provide more comprehensive coverage and semantic richness. This strategy ensures that user queries can match either question content or answer content, increasing retrieval accuracy and flexibility. The complete document is stored as payload metadata, preserving all original fields for filtering and providing rich context for results.
```python
for idx, document in enumerate(documents):
    # Combine question and answer for better retrieval
    combined_text = document["question"] + " " + document["text"]
    
    point = {
        "id": idx,
        "vector": {
            "text": combined_text,
            "model": model_handle
        },
        "payload": document  # Complete document as metadata
    }
```

````python
points = []

for i, doc in enumerate(documents):
    text = doc['question'] + ' ' + doc['text']
    vector = models.Document(text=text, model=model_handle)
    point = models.PointStruct(
        id=i,
        vector=vector,
        payload=doc
    )
    points.append(point)

qd_client.upsert(
    collection_name=collection_name,
    points=points
)
````

The vector search function encapsulates query embedding, filtering logic, and result processing into a clean, reusable interface. The implementation converts user questions into the same vector space used during indexing, applies configurable course filters for targeted results, and processes the returned points to extract human-readable content. This approach ensures consistent semantic comparison accuracy while providing flexible filtering capabilities.

````python
question = 'I just discovered the course. Can I still join it?'

def vector_search(question):
    print('vector_search is used')
    course = 'data-engineering-zoomcamp'
    query_points = qd_client.query_points(
        collection_name=collection_name,
        query=models.Document(
            text=question,
            model=model_handle 
        ),
        query_filter=models.Filter( 
            must=[
                models.FieldCondition(
                    key="course",
                    match=models.MatchValue(value=course)
                )
            ]
        ),
        limit=5,
        with_payload=True
    )
    results = []
    for point in query_points.points:
        results.append(point.payload)
    return results

def rag(query):
    search_results = vector_search(query)
    prompt = build_prompt(query, search_results)
    answer = llm(prompt)
    return answer
````

Testing the search function involves running sample queries and analyzing result quality to ensure semantic matching effectiveness.
